Tips Dataset: EDA Project
This dataset has 244 restaurant bills. Each row has the bill, the tip, the gender of the person who paid, smoker (Yes/No), day, time (Lunch/Dinner) and table size.

**Questions:**
1. How big is a normal tip, in dollars and in percent of the bill?
2. Does a bigger bill mean a bigger tip?
3. Does gender change how much people tip?
4. Do smokers tip differently from non-smokers?
5. Are there unusually big tips? Are they mistakes?

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("tips.csv")
df.head()
```

```python
print(df.shape)
```

```python
df.info()
```

```python
df.describe().round(1)
```

```python
print("Missing values:")
print(df.isna().sum())
```

No missing values, so nothing was dropped or filled.

```python
df = df.rename(columns={"sex": "gender"})

print(df.columns)
print(df["gender"].value_counts())
```

```python
print("Maximum tip:", df["tip"].max())
print("Minimum tip:", df["tip"].min())
```

```python
print("Maximum tip (Day Wise)", df.groupby("day")["tip"].max())
print("\nMinimum tip (Day Wise)", df.groupby("day")["tip"].min())
```

```python
print("Average tip by gender",df.groupby("gender")["tip"].mean().round(1))
print("Maximum tip by smoker",df.groupby("smoker")["tip"].max().round(1))
print("Maximum tip by time",df.groupby("time")["tip"].max().round(1))
```

```python
print(df["gender"].value_counts())
print(df["day"].value_counts())
print(df["time"].value_counts())
```

```python
df["tip_pct"] = (df["tip"] / df["total_bill"]) * 100
print("Average tip percentage", df["tip_pct"].mean().round(2), "%")
print("Median tip percentage", df["tip_pct"].median().round(2), "%")
print("Median tip amount", df["tip"].median().round(2))
print("Average tip amount", df["tip"].mean().round(2))
print("Average bill amount", df["total_bill"].mean().round(2))
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="size")
plt.title("Number of Tables by Size")
plt.xlabel("Size")
plt.ylabel("Number of tables")
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="tip")
plt.title("Distribution and Outliers of Tip Amount")
plt.xlabel("Tip amount")
plt.savefig("tip_outliers.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="tip_pct", bins=25)
plt.title("Distribution of Tip Percentage")
plt.xlabel("Tip as percent of bill")
plt.savefig("tip_pct_distribution.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
q1 = df["tip"].quantile(0.25)
q3 = df["tip"].quantile(0.75)
iqr = q3 - q1
high = q3 + 1.5 * iqr
low = q1 - 1.5 * iqr

print("Q1:", q1, " Q3:", q3, " IQR:", round(iqr, 1))
print("Above", round(high, 1), "is an outlier")
print("Below", round(low, 1), "is an outlier")

outliers = df[(df["tip"] < low) | (df["tip"] > high)]
print("Number of outliers:", len(outliers))
outliers.sort_values("tip", ascending=False).head(9)
```

The above are real big bills, not mistakes, so we keep them

```python
cols = ["tip", "total_bill", "size", "tip_pct"]
corr = df[cols].corr()
plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Between Tip, Bill, Size and Tip Percentage")
plt.savefig("correlation_heatmap.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="total_bill", y="tip")
plt.title("Bill vs Tip")
plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.savefig("bill_vs_tip.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
print(df.groupby("gender")["tip_pct"].mean().round(2))
print(df.groupby("smoker")["tip_pct"].mean().round(2))
print(df.groupby(["gender", "smoker"])["tip_pct"].mean().round(2))

plt.figure(figsize=(7, 5))
sns.barplot(data=df, x="gender", y="tip_pct", hue="smoker", errorbar=None)
plt.title("Average Tip Percentage by Gender and Smoking Status")
plt.xlabel("Gender")
plt.ylabel("Average tip %")
plt.savefig("tip_pct_gender_smoker.png", dpi=150, bbox_inches="tight")
plt.show()
```

# Findings

1. **Tip size:** The average tip is about **16%** of the bill. People tip a fairly steady share, whatever the bill size. (Chart: `tip_pct_gender_smoke.png`)

2. **Bill and tip:** The correlation between bill and tip is **0.68**, the strongest relation in the data. Bigger bills give bigger tips in dollars. (Charts: `bill_vs_tip.png`, `correlation_heatmap.png`)

3. **Gender and smoking:** Average tip % was **16.65% for women and 15.77% for men**, and **16.32% for smokers and 15.93% for non-smokers**. The differences are small, so these columns hardly matter. (Chart: `tip_pct_gender_smoker.png`)

4. **Outliers:** **9** tips are above the upper fence of **$5.9**. They come from real big bills, not mistakes, so I kept them. (Chart: `tip_outliers.png`)

# Conclusion

1. **How big is a normal tip?** About 16% of the bill.
2. **Does a bigger bill mean a bigger tip?** Yes, the correlation is 0.68.
3. **Does gender change tipping?** Almost not at all.
4. **Do smokers tip differently?** Almost not at all.
5. **Are the big tips mistakes?** No, there are 9 of them, they are real, and I kept them.
