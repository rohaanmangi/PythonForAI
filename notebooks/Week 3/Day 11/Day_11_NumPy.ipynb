# Hour 1: Arrays, Shape and Indexing

## CONCEPT: Why do we need NumPy?

- A Python **list** can hold numbers. But for maths, we need a loop every time.
- NumPy gives us the **array**. An array is a grid of numbers made for fast maths.
- Pandas, scikit-learn and PyTorch all keep their numbers in NumPy arrays.
- Arrays are fast, because NumPy does the loop inside fast C code, not in Python.

### EXAMPLE: Import NumPy

*We call it `np`. Everyone in the world uses this short name.*

````python
import numpy as np
````

### EXAMPLE: A list cannot do maths on all numbers

*We want to add 100 to every price.*

````python
prices_list = [1200, 850, 400, 3500]
print(prices_list + 100)
````

That is an error. Now let's try `*`:

````python
print(prices_list * 2)
````

We wanted to double every price. But the list was **repeated** two times.

### EXAMPLE: The same maths with an array

````python
prices = np.array([1200, 850, 400, 3500])
print(prices + 100)
````

````python
print(prices * 2)
````

No loop. NumPy does the maths on **every number**. This is called a **vectorized** operation.

## CONCEPT: Arrays are fast

- A Python loop works on one number at a time.
- NumPy works on all numbers together.
- AI needs billions of calculations. This is why AI uses NumPy.

**Before we run the speed test:** the list side uses a *list comprehension*. It is a short way to write a loop that makes a list.

Normal loop (4 lines):
```python
result = []
for x in numbers:
    result.append(x * 2)
```

List comprehension (1 line, same answer):
```python
result = [x * 2 for x in numbers]
```

How to read it:
```
[ x * 2   for x in numbers ]
  ↑         ↑
  put this  take each x
  in the    from numbers
  new list
```
In words: "For each `x` in `numbers`, put `x * 2` in the new list."

### EXAMPLE: Speed test: 1 million numbers

*`1_000_000` is the same as `1000000`. The `_` only helps us read it.*

````python
import time

numbers = list(range(1_000_000))
arr = np.array(numbers)

start = time.time()
result = [x * 2 for x in numbers]
print("list  :", time.time() - start, "seconds")

start = time.time()
result = arr * 2
print("array :", time.time() - start, "seconds")
````

## CONCEPT: Dimensions and shape

- A **1-D** array is one row of numbers: `np.array([1, 2, 3])`.
- A **2-D** array is a table (a list of lists): `np.array([[1, 2], [3, 4]])`.
- `.ndim` = how many directions (dimensions).
- `.shape` = size in each direction. For 2-D: `(rows, columns)`.
- `.size` = how many numbers in total.
- `.dtype` = the type of the numbers (`int64`, `float64`).

### EXAMPLE: Look inside an array

*The same 4 things work on any array.*

````python
grid = np.array([[1, 2, 3],
                 [4, 5, 6]])
print(grid)
````

````python
print(grid.ndim)     # 2
print(grid.shape)    # (2, 3) -> 2 rows, 3 columns
print(grid.size)     # 6
print(grid.dtype)    # int64
````

**Tip:** When something breaks, first print the `shape`. Most NumPy errors are shape errors.

## CONCEPT: Data type (dtype)

- An array keeps **only one type** of data.
- If you put a decimal into a whole-number array, the decimal part is **cut off**. There is no warning.
- If you mix text and numbers, **everything becomes text**.
- If you need decimals, make the array with `dtype=float`.

### EXAMPLE: The dtype problem

````python
marks = np.array([80, 72, 91])
print(marks.dtype)
````

**WRONG ON PURPOSE.** We put `85.7` in a whole-number array:

````python
marks[0] = 85.7
print(marks)
````

We wrote 85.7. We got 85. The `.7` was lost.

Mixing text and numbers:

````python
mixed = np.array([1, 2, "3"])
print(mixed, mixed.dtype)
````

**Fix:** use `dtype=float`.

````python
marks = np.array([80, 72, 91], dtype=float)
marks[0] = 85.7
print(marks)
````

## CONCEPT: Indexing: get one value

- A 1-D array works like a list: `arr[0]`, `arr[-1]`.
- A 2-D array uses **one bracket with a comma**: `grid[row, col]`.
- Counting starts from **0**.
- Negative numbers work too: `grid[-1, -1]` is the last row, last column.

### EXAMPLE: Indexing a 2-D array

````python
marks = np.array([[80, 72, 91],    # Ali
                  [65, 88, 70],    # Sara
                  [95, 60, 84]])   # Hassan
#                  Eng Maths Sci
print(marks)
````

````python
print(marks[0, 0])     # row 0, column 0 -> Ali's English
print(marks[1, 2])     # Sara's Science
print(marks[-1, -1])   # last row, last column -> Hassan's Science
````

````python
print(marks[0])        # only the row -> all of Ali's marks
````

## CONCEPT: A slice is a view, not a copy

- With a **list**, a slice is a new copy.
- With a **NumPy array**, a slice **points to the same data**.
- If you change the slice, the original changes too.
- If you want a separate copy, use `.copy()`.

### EXAMPLE: The view problem

**WRONG ON PURPOSE.**

````python
first_row = marks[0]
print(first_row)
first_row[0] = 0
print(marks)
````

We changed `first_row`. But Ali's English in `marks` is also 0 now.

Now the safe way, with `.copy()`:

````python
marks[0, 0] = 80              # put it back
safe = marks[0].copy()
safe[0] = 0
print(marks[0])               # the original is safe
print(safe)
````

## CONCEPT: Filters: pick values with a condition

- When you compare an array, you get a **True/False** array. It is called a mask.
- Put the mask inside `[ ]`. NumPy keeps only the `True` values.
- `True` counts as 1, so `.sum()` counts how many are `True`.
- For "and", "or", "not" on arrays, use `&`, `|`, `~`. Put brackets around each condition.

### EXAMPLE: Filter without a loop

````python
scores = np.array([45, 82, 67, 91, 38, 74])
print(scores >= 50)
````

````python
print(scores[scores >= 50])             # only the passed students
````

````python
print((scores >= 50).sum(), "passed")   # count the True values
````

````python
scores[scores < 50] = 50                # grace marks: raise every fail to 50
print(scores)
````

**BREAKS ON PURPOSE.** We want marks between 50 and 80. Let's use `and`:

````python
scores[scores > 50 and scores < 80]
````

`and` can only check one True/False. We gave it six. Use `&` with brackets:

````python
print(scores[(scores > 50) | (scores < 80)])
````

## CONCEPT: Totals: sum, mean, max, min

- These functions turn many numbers into **one answer**.
- `arr.sum()` = total, `arr.mean()` = average, `arr.max()` = biggest, `arr.min()` = smallest.

### EXAMPLE: Totals of the whole array

````python
marks = np.array([[80, 72, 91],
                  [65, 88, 70],
                  [95, 60, 84]])

print(marks.sum())     # total of all numbers
print(marks.mean())    # average
print(marks.max())     # biggest
print(marks.min())     # smallest
````

## CONCEPT: axis: rows or columns?

- Most times we want an answer for **each row** or **each column**.
- `axis=0` goes **down** → one answer for each **column** (each subject).
- `axis=1` goes **across** → one answer for each **row** (each student).
- `argmax()` gives the **position** of the biggest number, not the number.

### EXAMPLE: Using axis

*Rows = students, columns = Eng, Maths, Sci.*

````python
print(marks.mean(axis=0))    # average of each subject
````

````python
print(marks.sum(axis=1))     # total of each student
````

````python
print(marks.max(axis=0))     # best mark in each subject
````

````python
print(marks.argmax(axis=1))  # which student is best in each subject
````

# Hour 3: Broadcasting and Practice

## CONCEPT: Broadcasting

- You already used it: `prices + 100` added one number to every price.
- NumPy can also use a **small array** with a **big array**. It repeats the small one to fit.
- The rule: the sizes must be the **same**, or one of them must be **1**.
- When you get a broadcast error, `print` both shapes and compare them.

### EXAMPLE: Add a bonus to Maths only

*3 subjects, so the bonus has 3 numbers.*

````python
bonus = np.array([0, 5, 0])     # Eng, Maths, Sci
print(marks + bonus)
````

NumPy added `[0, 5, 0]` to **every row**:

```
[[80 72 91]     [0 5 0]
 [65 88 70]  +  [0 5 0]
 [95 60 84]]    [0 5 0]
```

**BREAKS ON PURPOSE.** Only 2 bonus numbers for 3 subjects:

````python
bad_bonus = np.array([5, 10])
marks + bad_bonus
````

# Small example: electricity bill

A meter reads the units used by 3 homes on 4 days. Each row is one home.

````python
units = np.array([[5, 6, 4, 7],      # Home 1
                  [10, 12, 9, 11],   # Home 2
                  [3, 2, 4, 3]])     # Home 3
print(units.shape)
````

**Question 1:** How many units did each home use in total?

````python
total = units.sum(axis=1)
print(total)
````

**Question 2:** Each unit costs Rs 40. What is each home's bill?

````python
bill = total * 40
print(bill)
````

**Question 3:** Which home pays the most?

````python
print("Home number:", bill.argmax() + 1)    # +1 because counting starts from 0
````

**Question 4:** Which day had the most use in total?

````python
per_day = units.sum(axis=0)
print(per_day)
print("Busiest day:", per_day.argmax() + 1)
````

We answered all of these with no loops. Real electricity companies do this for millions of homes. The code is the same, only the array is bigger.

# Key Takeaways

- A NumPy array is a grid of numbers of **one type**, made for fast maths.
- `.shape`, `.ndim`, `.size` and `.dtype` tell you about an array.
- Use `arr[row, col]` to get one value. Use `:` to get many.
- A slice is a **view**. Use `.copy()` for a real copy.
- Use `&`, `|`, `~` (not `and`, `or`, `not`) for filters.
- `axis=0` goes down (columns). `axis=1` goes across (rows).
- Vectorized maths and broadcasting replace loops. This is why NumPy is fast.
